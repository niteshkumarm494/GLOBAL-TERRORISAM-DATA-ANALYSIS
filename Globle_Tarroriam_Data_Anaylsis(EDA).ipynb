# **Counts and print only null cloumns and percentage of missing values in each column**

````python
def count_null_percentage(df):
    """
    Counts the percentage of missing values in each column and prints
    only those columns that have missing values.
    Also prints the total missing percentage for the entire dataset.
    """
    total_rows = len(df)  # Get total number of rows
    total_cells = df.size  # Total number of cells in the DataFrame
    null_counts = df.isnull().sum()  # Count missing values per column
    null_percentage = (null_counts / total_rows) * 100  # Calculate column-wise missing percentage

    # Create a DataFrame with only columns that have missing values
    null_report = pd.DataFrame({
        "Column Name": df.columns,
        "Missing Count": null_counts,
        "Missing Percentage": null_percentage
    })

    # Filter to keep only columns where Missing Count > 0
    null_report = null_report[null_report["Missing Count"] > 0].reset_index(drop=True)

    # Calculate total missing values for the entire dataset
    total_missing_values = df.isnull().sum().sum()  # Sum of all missing values
    total_missing_percentage = (total_missing_values / total_cells) * 100  # Missing percentage of entire dataset

    # Print the report
    print("\n Columns with Missing Values:\n")
    print(null_report)
    print("\n Total Missing Values in Dataset:", total_missing_values)
    print(f" Total Dataset Missing Percentage: {total_missing_percentage:.2f}%")

    return null_report

# Example Usage
# Load dataset
print("Total Raw :",df.shape[0])
print("TOtal Columns: ",df.shape[1])
null_report = count_null_percentage(df).iloc  # Apply function
````

# **Drops columns where more than '20%' percentage of  Total rows have null values**

````python
def drop_high_null_columns(df, threshold=0.2):
    """
    Drops columns where more than 'threshold' percentage of rows have null values.
    percentage (0.2 = 20%) of missing values to drop a column.
    """
    total_rows = len(df)  # Get total number of rows
    missing_counts = df.isnull().sum()  # Count missing values per column
    missing_ratio = missing_counts / total_rows  # Get ratio of missing values

    # Select columns where the missing ratio is greater than threshold
    cols_to_drop = missing_ratio[missing_ratio > threshold].index

    print(f"Dropping {len(cols_to_drop)} columns with more than {threshold * 100}% missing values.\n")

    return df.drop(columns=cols_to_drop)  # Drop selected columns

# Example Usage

df_cleaned = drop_high_null_columns(df)  # Apply the function
print("Updated DataFrame Shape:", df_cleaned.shape)  # Print new shape
````

#**Filling missing values with the appropriate method**

````python
def fill_missing_values(df, threshold=10):
    """
    Function to handle missing values in a dataset.
    - Categorical values -> "Unknown" (if missing ≤ 10%)
    - Binary values -> Mode (most common value)
    - Numerical values -> Mean (latitude/longitude) or Median (other numerical)
    - Casualties -> Fill with 0 (assuming missing = no casualties)
    """
    df_cleaned = df.copy()  # Create a copy of the DataFrame to avoid modifying the original

    # Calculate missing percentage for each column
    missing_percent = (df_cleaned.isnull().sum() / len(df_cleaned)) * 100

    # Categorical Columns: Fill with "Unknown" if missing values ≤ 10%
    categorical_columns = ["provstate", "city", "targsubtype1_txt", "target1", "natlty1_txt", "weapsubtype1_txt"]
    for col in categorical_columns:
        if col in df_cleaned.columns and missing_percent[col] <= threshold:
            df_cleaned[col] = df_cleaned[col].fillna("Unknown")

    # Binary Columns: Fill with mode (most frequent value) if missing values ≤ 10%
    binary_columns = ["doubtterr", "multiple", "guncertain1", "ishostkid"]
    for col in binary_columns:
        if col in df_cleaned.columns and missing_percent[col] <= threshold:
            df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mode()[0])

    # Latitude & Longitude: Fill with mean if missing values ≤ 10%
    geo_columns = ["latitude", "longitude"]
    for col in geo_columns:
        if col in df_cleaned.columns and missing_percent[col] <= threshold:
            df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mean())

    # Casualties: Fill with 0 (assuming missing = no casualties)
    casualty_columns = ["nkill", "nwound","natlty1"]
    for col in casualty_columns:
        if col in df_cleaned.columns and missing_percent[col] <= threshold:
            df_cleaned[col] = df_cleaned[col].fillna(0)

    # Numerical Columns: Fill with median if missing values ≤ 10%
    numerical_columns = ["specificity"]
    for col in numerical_columns:
        if col in df_cleaned.columns and missing_percent[col] <= threshold:
            df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].median())

    return df_cleaned

df_cleaned = fill_missing_values(df_cleaned)
final = count_null_percentage(df_cleaned)
````

````python
# Chart - 11 visualization code

# Interactive function to show most affected countries in a selected region
@interact
def attack_region(region_txt=list(df_copy['region_txt'].unique())):
    # Filter data for the selected region
    attack = df_copy[df_copy['region_txt'] == region_txt]

    # Count attacks per country in the selected region
    country_counts = attack['country_txt'].value_counts().reset_index()
    country_counts.columns = ["Country", "Number of Attacks"]

    # Set figure size
    plt.figure(figsize=(16, 8))

    # Create a horizontal bar chart
    ax = sns.barplot(y=country_counts["Country"], x=country_counts["Number of Attacks"], palette="plasma")

    # **Adding value labels on bars**
    for index, value in enumerate(country_counts["Number of Attacks"]):
        plt.text(value + 5, index, str(value), fontsize=12, va="center", color='black', fontweight='bold')

    # Titles and labels
    plt.title(f"Most Affected Countries in {region_txt}", fontsize=14, fontweight="bold")
    plt.xlabel("Number of Attacks", fontsize=12)
    plt.ylabel("Countries", fontsize=12)

    # Show plot
    plt.show()
````
